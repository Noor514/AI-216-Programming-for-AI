```python
import sqlite3
import requests
from bs4 import BeautifulSoup
quotes_data = []

for page in range(1, 4):
    url = f"http://quotes.toscrape.com/page/{page}/"
    response = requests.get(url)
    soup = BeautifulSoup(response.text, "html.parser")

    quotes = soup.find_all("div", class_="quote")

    for q in quotes:
        text = q.find("span", class_="text").get_text()
        author = q.find("small", class_="author").get_text()

        quotes_data.append((text, author))

conn = sqlite3.connect("quotes.db")
cursor = conn.cursor()
cursor.executemany(
    "INSERT INTO quotes (quote, author) VALUES (?, ?)",
    quotes_data
)

conn.commit()
print(f"{len(quotes_data)} quotes inserted successfully!")
conn.close()
```

```python
import sqlite3
conn = sqlite3.connect("quotes.db")
cursor = conn.cursor()
cursor.execute("SELECT author FROM quotes LIMIT 1")
result = cursor.fetchone()

if result:
    author_name = result[0]
    print("Selected author:", author_name)
    cursor.execute(
        "UPDATE quotes SET author = UPPER(author) WHERE author = ?",
        (author_name,)
    )

    conn.commit()

    print(f"Author '{author_name}' updated to uppercase.")

else:
    print("No data found in database.")

conn.close()
```

```python
import sqlite3

conn = sqlite3.connect("quotes.db")
cursor = conn.cursor()

cursor.execute("SELECT COUNT(*) FROM quotes")
before_count = cursor.fetchone()[0]

cursor.execute("DELETE FROM quotes WHERE LENGTH(quote) < 30")

conn.commit()

cursor.execute("SELECT COUNT(*) FROM quotes")
after_count = cursor.fetchone()[0]

deleted = before_count - after_count
print(f"{deleted} quotes deleted (less than 30 characters).")
conn.close()
```

```python
import requests
import sqlite3
from bs4 import BeautifulSoup

conn = sqlite3.connect("quotes.db")
cursor = conn.cursor()

cursor.execute("""
CREATE TABLE IF NOT EXISTS quotes (
    id INTEGER PRIMARY KEY,
    quote TEXT,
    author TEXT
)
""")

cursor.execute("DELETE FROM quotes")

quotes_data = []

for page in range(1, 4):
    url = f"http://quotes.toscrape.com/page/{page}/"
    response = requests.get(url)
    soup = BeautifulSoup(response.text, "html.parser")

    quotes = soup.find_all("div", class_="quote")

    for q in quotes:
        quote_text = q.find("span", class_="text").get_text()
        author = q.find("small", class_="author").get_text()
        quotes_data.append((quote_text, author))

cursor.executemany(
    "INSERT INTO quotes (quote, author) VALUES (?, ?)",
    quotes_data
)

conn.commit()


cursor.execute("SELECT COUNT(*) FROM quotes")
total_quotes = cursor.fetchone()[0]

cursor.execute("SELECT COUNT(DISTINCT author) FROM quotes")
unique_authors = cursor.fetchone()[0]

cursor.execute("""
SELECT author, COUNT(*) as cnt
FROM quotes
GROUP BY author
ORDER BY cnt DESC
LIMIT 1
""")
top_author = cursor.fetchone()

print("===== QUOTES REPORT =====")
print("Total quotes stored:", total_quotes)
print("Total unique authors:", unique_authors)
print("Author with most quotes:", top_author[0], f"({top_author[1]} quotes)")

conn.close()
```
