```python
prices = [1200, 850, 4300, 2999, 1500, 850]
total_products=len(prices)
print("total products:",total_products)
average=sum(prices)/total_products
print("average:",average)
products_above=[]
for price in prices:
  if price > 2000:
     products_above.append(price)
  print("product above 2000:",products_above)
  prices.sort(reverse=True)
  print("price in descending order:",prices)
```

```python
python_students = {"Ali", "Sara", "Ahmed", "Zain"}
ml_students = {"Sara", "Hassan", "Ali", "Fatima"}
python_course=python_students.intersection(ml_students)
print("Students in both courses:", python_course)
only_python=python_students.difference(ml_students)
print("Students only in Python:", only_python)
master_set = python_students.union(ml_students)
print("Master student set:", master_set)
ml_students.remove("Ali")
print("Updated ML students:", ml_students)
```

```python
employees = {
    "E01": {"name": "Ali", "salary": 60000},
    "E02": {"name": "Sara", "salary": 75000},
    "E03": {"name": "Ahmed", "salary": 50000}
}
print("employee IDs:",list(employees.keys()))
total_payroll= sum(emp["salary"] for emp in employees.values())
print("total payroll:,total_payroll")
high_earners=[]
for emp_id,emp_info in employees.items():
    if emp_info["salary"]> 60000:
        high_earners.append(emp_info["name"])
print("Employess earning above 60000:", high_earners)

employees["E04"] = {"name": "Zain", "salary": 65000}
print("Updated employee records:", employees)
```

```python

sales_data = {
    "electronics": [
        ("Laptop", 120000, 3),
        ("Mouse", 1500, 10)
    ],
    "clothing": [
        ("Shirt", 2500, 5),
        ("Shoes", 5000, 2)
    ]
}
category_revenue = {}
unique_products = set()

for category, items in sales_data.items():
    total = 0
    for product_name, price, quantity in items:
        total += price * quantity
        unique_products.add(product_name)
    category_revenue[category] = total

print("Revenue Report:")
for category, revenue in category_revenue.items():
    print(f"{category.capitalize()}: Rs {revenue}")

print("\nUnique products across all categories:", unique_products)
```
